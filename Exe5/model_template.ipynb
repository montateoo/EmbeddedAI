```python
import numpy as np

fall_dir = Path("fall")
non_fall_dir = Path("non_fall")
assert fall_dir.exists(), f"Fall folder {fall_dir} not found."
assert non_fall_dir.exists(), f"Non-fall folder {non_fall_dir} not found."

def load_npy_files(folder: Path):
    data = {}
    for file_path in sorted(folder.glob("*.npy")):
        arr = np.load(file_path)
        data[file_path.name] = arr
    return data

fall_data = load_npy_files(fall_dir)
non_fall_data = load_npy_files(non_fall_dir)

print(f"Loaded {len(fall_data)} fall arrays and {len(non_fall_data)} non-fall arrays.")
print("Sample shapes from fall:")
for name, arr in list(fall_data.items())[:5]:
    print(f"  {name}: {arr.shape}")

print("Sample shapes from non-fall:")
for name, arr in list(non_fall_data.items())[:5]:
    print(f"  {name}: {arr.shape}")
```
